# コラム: 計算量のオーダー記法

古典・量子を問わず，アルゴリズムの性能を評価する際には，実行される基本演算の回数や必要なメモリの大きさを見積もることが重要である。そのためのおおまかな評価方法として，オーダー記法が用いられる。

解きたい問題のサイズを $n$ として，2つの関数 $f(n), g(n)$ が与えられたとする。このとき，$n$ が十分大きい領域で $f(n)$ が $g(n)$ に比例するか，あるいはそれ以下の大きさで抑えられるとき，

$$
f(n) = O(g(n))
$$

と書く。実用上は $g(n)$ として $n^2$，$e^n$，$\log n$ のような代表的な増え方を考えることが多い。たとえば $f(n)=2n^3+4n+3$ は，大きな $n$ では $n^3$ の項が支配的になるので $O(n^3)$ と書ける。

注意点として，$O$ 記法は「同じオーダー」ではなく「それ以下で抑えられる」を表すため，$3n^2 = O(n^3)$ も正しい。また，これはあくまで $n$ が十分大きいときの振る舞いを示すだけなので，実際の問題サイズでどちらが大きいかをそのまま保証するものではない。
