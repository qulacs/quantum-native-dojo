# コラム: 量子複製不可能定理

量子情報が古典情報と大きく異なる点の1つに、未知の量子状態をそのまま完全複製することはできない、という事実がある。これを **量子複製不可能定理** と呼ぶ。古典ビットであれば 0 や 1 をそのままコピーできるが、量子状態では一般の重ね合わせ状態に対して同じことはできない。

定理の主張は、任意の状態 $|\psi\rangle$ に対して

$$
U(|\psi\rangle \otimes |0\rangle) = |\psi\rangle \otimes |\psi\rangle
$$

を満たす1つのユニタリー演算子 $U$ は存在しない、というものである。つまり、入力状態が何であっても同じ回路で完全コピーを作ることはできない。

証明の考え方はシンプルである。もしそのようなユニタリー $U$ が存在すると仮定し、任意の2状態 $|\psi\rangle, |\phi\rangle$ に対して

$$
U(|\psi\rangle \otimes |0\rangle) = |\psi\rangle \otimes |\psi\rangle,
\qquad
U(|\phi\rangle \otimes |0\rangle) = |\phi\rangle \otimes |\phi\rangle
$$

が成り立つとする。この2式の内積を取ると、ユニタリー性 $U^\dagger U = I$ から

$$
\langle \psi | \phi \rangle = (\langle \psi | \phi \rangle)^2
$$

を得る。したがって $\langle \psi | \phi \rangle$ は 0 または 1 でなければならない。しかし実際には、異なる量子状態の内積はその間のさまざまな値を取りうる。よって仮定が矛盾し、そのような万能複製器は存在しない。

この定理は、量子通信や量子暗号の安全性を支える基本原理でもある。たとえば盗聴者が未知の量子状態をこっそり複製してから測定する、という古典的な発想は量子系では通用しない。量子情報科学のさまざまな特徴は、このような「コピーできなさ」に根ざしている。
