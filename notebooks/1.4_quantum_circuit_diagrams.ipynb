## 1-4. 量子回路図

ここまでに学んだ量子ビットの状態や量子ゲートは、実際のアルゴリズムでは量子回路図としてまとめて表現することが多い。量子回路図は、どの量子ビットにどの順番で操作を加えるかを視覚的に表したものであり、量子情報科学の共通言語になっている。

量子回路図では、各量子ビットは横方向の線で表される。左端は初期状態、右端は最終的な測定や出力に対応し、回路は **左から右へ** 読む。したがって、横線上に並んだゲートを順に適用していくことで、量子状態の時間発展や計算の流れを追うことができる。

量子回路図の主な構成要素は次のとおりである。

- 量子ビット: 1本の横線が1つの量子ビットを表す。
- 量子ゲート: 線上の箱や記号が、その量子ビットに作用するゲートを表す。
- 多体系ゲート: 複数の線にまたがる箱や、制御記号と標的記号を結ぶ縦線で表現される。
- 測定: 右端のメーター記号などで表され、量子状態を古典情報へ変換する操作に対応する。

たとえばアダマールゲート $H$ を1量子ビットに作用させ、その後に CNOT ゲートを適用すると、重ね合わせやエンタングルメントを作る回路を簡潔に描ける。複数の量子ゲートを数式だけで追うと煩雑になりやすいが、回路図を使うとどの量子ビットが制御側で、どこに測定が入るのかが一目で分かる。
