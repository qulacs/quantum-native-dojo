## 2-1. 量子アルゴリズムの分類

現在発明・発見されている量子アルゴリズムは，実用上の問題に対する実現可能性の観点から，NISQ アルゴリズム，Early-FTQC アルゴリズム，FTQC アルゴリズムという 3 つのグループに大別できる．ただしこれらの区別は絶対的なものではなく，解くべき問題の大きさや量子コンピュータハードウェアの進歩によって移り変わる点に注意が必要である。

### 量子コンピュータハードウェアの発展

2026 年現在で実現されている量子コンピュータは，量子誤り訂正機能をもたない NISQ デバイスが中心である。NISQ デバイスは数十個から 1000 個程度の量子ビットをもちうるが，計算途中のノイズの影響を強く受けるため，深い回路や長い計算には向かない。

その先に想定される段階が Early-FTQC である。これは誤り訂正機能を備えて NISQ より高精度・大規模な計算が可能になったものの，まだ大規模な素因数分解や本格的な量子化学計算を容易には実行できない黎明期の FTQC を指す。さらにその先に，多数の論理量子ビットを用いて複雑なアルゴリズムを安定に実行できる FTQC がある。

### アルゴリズムの分類

NISQ アルゴリズムは，少数の量子ビットと浅い回路で実行できることを重視したアルゴリズム群であり，量子古典ハイブリッド計算と相性がよい。代表例としては VQE，QAOA，QCL などが挙げられる。

Early-FTQC アルゴリズムは，誤り訂正による高精度化を前提としつつも，まだ補助量子ビット数や計算規模に強い制約がある時代を見据えたアルゴリズム群である。NISQ では厳しいが，完全成熟した FTQC を待たなくても価値を発揮しうる計算がここに含まれる。

FTQC アルゴリズムは，多数の論理量子ビットと高精度な誤り訂正が前提となる本格的な量子アルゴリズム群である。ショアのアルゴリズム，量子位相推定，HHL アルゴリズムなどが典型例であり，量子コンピュータの指数関数的な優位性が期待される領域と深く結びついている。

この章で学ぶアダマールテスト，量子フーリエ変換，量子位相推定は，FTQC 時代の量子アルゴリズムを理解するための基礎にあたる。一方で，後続章で扱う VQE や QAOA は NISQ 的な発想にもつながっており，本節の分類は以後の章全体を見通すための地図として役立つ。
