# コラム: 量子計算のユニバーサルゲートセット

古典計算では、AND や NOT のような少数の論理ゲートを組み合わせることで任意の論理回路を構成できる。同じように量子計算でも、限られた種類の量子ゲートの組合せから任意の量子演算を近似できる。このようなゲートの集まりを **ユニバーサルゲートセット** と呼ぶ。

量子計算では、すべての量子演算を有限個の基本ゲートだけで厳密に表すことは一般にはできない。しかし、適切なゲート集合を用いれば、任意のユニタリー演算を任意の精度で近似できることが知られている。代表的な例としては、1量子ビットゲートに加えて CNOT のような2量子ビットゲートを含む集合がある。

実際によく挙げられる組合せの1つが $\{H, T, \mathrm{CNOT}\}$ である。ここで $H$ は重ね合わせを作るアダマールゲート、$T$ は位相を細かく調整するゲート、CNOT は量子ビット同士を相互作用させてエンタングルメントを生成するゲートである。このように、1量子ビットの回転を担うゲートと、量子ビット間の結合を担うゲートを組み合わせることで、表現力が一気に高まる。

ユニバーサルゲートセットの考え方は、理論的な量子計算の基礎であると同時に、実機実装にも直結している。ハードウェアごとに実装しやすいネイティブゲートは異なるが、最終的にはそれらを基本ゲートに分解して量子回路を実行することになる。したがって、どのゲート集合を基本とみなすかは、アルゴリズム設計とコンパイルの両方に深く関わる。
