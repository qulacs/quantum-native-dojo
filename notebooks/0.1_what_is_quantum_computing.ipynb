# 0-1. 量子コンピュータとは

量子コンピュータは、私たちが普段使っている古典コンピュータとは異なる原理で動作する計算機である。この節では、まず量子という概念そのものを振り返り、そのうえで量子コンピュータがなぜ期待されているのか、そして実際の計算がどのような流れで進むのかを概観する。

## 0.1.1 量子という概念

量子論の出発点としてよく挙げられるのが、1900年にマックス・プランクが与えた熱放射の理論である。高温になった物質が放出する電磁波のスペクトルを説明するために、プランクはエネルギーが連続的ではなく、離散的な単位でやり取りされると仮定した。この考え方は、エネルギー量子

$$
E = h\nu
$$

という式で表される。ここで $h$ はプランク定数、$\nu$ は振動数である。この仮定は当初は数学的な工夫のように見えたが、後の量子論の基礎になった。

1905年にはアルベルト・アインシュタインがこの考えを光電効果の説明に使い、光が波であるだけでなく粒子としても振る舞うことを示した。さらに1924年にはルイ・ド・ブロイが電子などの粒子にも波の性質があると提案し、1926年にはエルヴィン・シュレーディンガーが波動方程式を定式化した。これにより、量子力学では粒子の状態は波動関数によって記述され、その絶対値の二乗が観測確率を与えるという描像が成立した。

量子力学の本質は、状態が重ね合わせとして存在できることにある。観測前の量子系は複数の可能性を同時に含んでおり、測定によってはじめて 1 つの結果に収縮する。この直感に反する性質を象徴的に示す思考実験がシュレーディンガーの猫である。

量子コンピュータで使う量子ビットは、この重ね合わせの性質を情報表現に利用する。古典ビットが 0 または 1 のどちらか一方しか取れないのに対し、量子ビットは 0 と 1 の重ね合わせ状態を取ることができる。さらに複数の量子ビットの間にはエンタングルメントと呼ばれる強い相関を作ることができ、量子干渉と組み合わせることで、特定の答えに対応する確率振幅を強めるような計算が可能になる。

つまり、量子コンピュータとは単に小さな物理現象を使う計算機ではなく、重ね合わせ、干渉、エンタングルメントといった量子的性質そのものを計算資源として活用する装置である。

## 0.1.2 量子コンピュータへの期待とその応用先

量子コンピュータが注目される背景には、現代の情報技術が直面している 2 つの限界がある。1 つは、古典コンピュータのハードウェア微細化が原子スケールに近づき、熱雑音や量子トンネル効果などの物理的制約が無視できなくなっていること。もう 1 つは、材料科学、量子化学、組合せ最適化のように、問題サイズが少し大きくなるだけで計算量が爆発的に増える課題が数多く存在することである。

たとえば、材料中の電子状態計算、分子の構造や反応経路の予測、物流ネットワークの最適化などでは、古典コンピュータでの探索空間が指数関数的に大きくなり、現実的な時間で解くのが困難になることがある。量子コンピュータは、量子ビットの重ね合わせによって膨大な状態を同時に表現し、さらに干渉を利用して有望な答えを強調することで、こうした課題に対して古典計算とは異なるアプローチを与える。

ただし、量子コンピュータは万能な高速化装置ではない。ショアのアルゴリズムやグローバーのアルゴリズムのように、数学的に優位性が示されている問題はあるが、あらゆる計算が速くなるわけではない。また、重ね合わせによって多数の状態を同時に表現できても、測定で得られる結果は最終的に 1 つだけである。そのため、正しい答えに高い確率で到達するように量子ゲート列を設計する必要がある。

量子コンピュータは、スマートフォンやノート PC をそのまま置き換える装置というより、従来の計算機では扱いづらい特定の課題に対して新しい道を切り開く専門的な計算機と理解するのがよい。

## 0.1.3 量子コンピュータの動作の流れ

量子計算の流れは、概念的には次の 3 段階に整理できる。

1. 初期状態の準備
2. 問題に対応する量子ゲート操作
3. 測定

まず、計算を始める前に量子ビットを既知の基準状態、典型的には $|0\rangle$ にそろえる。続いて、量子ゲートと呼ばれる操作を順に適用し、重ね合わせやエンタングルメントを利用しながら問題に対応する量子状態を作る。最後に測定を行うことで、古典的な 0 と 1 の列として結果を読み出す。

> 図 0.5 プレースホルダ
> 初期化・操作・測定からなる量子計算の流れを後で挿入

ここで重要なのは、量子計算は一般に一度の測定で必ず正解を返すとは限らない、という点である。測定結果は確率的であり、同じ回路を繰り返し実行して統計的に答えを推定することも多い。言い換えれば、量子コンピュータは一回で確定的な答えを返す装置というより、量子状態の設計と繰り返し測定によって解の輪郭を浮かび上がらせる計算機である。

この節で見た全体像は、第1章以降で扱う量子ビット、量子ゲート、量子アルゴリズムの基礎につながっていく。
