# コラム: 量子位相推定アルゴリズムの計算量と量子加速

量子位相推定アルゴリズムの計算量を考えるとき，まず古典コンピュータでは，$m$ 量子ビットに対応するユニタリ演算 $U$ は $2^m \times 2^m$ 行列であり，その固有値や固有ベクトルを一般に古典的に求めるには少なくとも指数関数的な計算量が必要になる。

一方，量子位相推定では，制御ユニタリゲート $\Lambda(U)$ を実装するコストを $C_{\Lambda(U)}$ とすると，補助量子ビット数を $n$ としたとき回路全体のゲート数はおおむね $O(2^n) \cdot C_{\Lambda(U)}$ で見積もられる。推定精度を $\varepsilon$ とすると $n = O(\log_2(1/\varepsilon))$ だから，計算量は $O(1/\varepsilon) \cdot C_{\Lambda(U)}$ とも書ける。

もし $U$ がシステムサイズ $m$ に対して多項式個のゲートで実装できるなら，量子位相推定全体の計算量も $\mathrm{poly}(m)/\varepsilon$ の形になり，古典計算の指数関数的コストに対して大きな優位性が期待できる。この意味で，量子位相推定は量子加速や量子優位性を示す有力な候補とみなされている。

ただし実際の性能は，回路のゲート数だけでは決まらない。初期状態が目的の固有状態にどれだけ重なっているかが小さいと，目的の固有値を得る確率が低くなり，回路を何度も繰り返し実行しなければならない。そのため，真の量子加速を評価するには，回路実装コストと成功確率の両方を見る必要がある。
