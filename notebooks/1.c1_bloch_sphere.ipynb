# コラム: ブロッホ球

1量子ビットの純粋状態は、規格化条件を満たす複素数 $\alpha, \beta$ を用いて $|\psi\rangle = \alpha |0\rangle + \beta |1\rangle$ と書ける。この表式は成分表示としては分かりやすいが、状態の全体像を直感的に捉えるには少し扱いにくい。そこで便利なのが、1量子ビット状態を球面上の点として表す **ブロッホ球** である。

グローバル位相を除くと、任意の1量子ビット純粋状態は

$$
|\psi\rangle = \cos\frac{\theta}{2}|0\rangle + e^{i\phi}\sin\frac{\theta}{2}|1\rangle
$$

と書ける。ここで $\theta$ と $\phi$ は球面座標に対応しており、状態は半径1の球の表面上の1点として表現される。北極が $|0\rangle$、南極が $|1\rangle$ に対応し、赤道上には $\frac{1}{\sqrt{2}}(|0\rangle + e^{i\phi}|1\rangle)$ のような等重みの重ね合わせ状態が並ぶ。

ブロッホ球の見方を使うと、量子ゲートの多くは球面上の回転として理解できる。たとえばパウリ $X, Y, Z$ 演算子はそれぞれ対応する軸まわりの回転と関係しており、アダマールゲートもブロッホ球上での向きの変換として解釈できる。成分計算だけでは見えにくい「位相の違い」や「どの軸のまわりに状態が変わるか」が、図形として把握しやすくなる。

ただし、ブロッホ球で素直に表せるのは1量子ビットの純粋状態に限られる。複数量子ビット系ではエンタングルメントが現れるため、同じような単純な幾何学図形だけで全状態を表すことはできない。それでも、1量子ビットの直感を身につける道具としてブロッホ球は非常に重要である。
