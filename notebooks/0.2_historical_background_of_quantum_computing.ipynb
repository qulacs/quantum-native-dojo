# 0-2. 量子コンピュータ開発の歴史的背景

量子コンピュータは、突然現れた技術ではない。古典コンピュータの発展とその限界、情報の物理学、量子力学の応用可能性、そして量子アルゴリズムの発見が積み重なって現在の研究分野が形づくられてきた。この節では、その流れを歴史に沿って整理する。

## ムーアの法則と古典計算の限界

1965年、ゴードン・ムーアは集積回路上のトランジスタ数が約 2 年ごとに倍増するという経験則を示した。ムーアの法則は、古典コンピュータの計算能力が指数関数的に向上してきたこと、そして同時にトランジスタのサイズが指数関数的に小さくなってきたことを意味する。

しかし、微細化が進みすぎると、単に性能が上がるだけでは済まなくなる。トランジスタが原子スケールに近づくにつれて、熱雑音や量子トンネル効果の影響が無視できなくなり、従来の古典的な描像のままでは高密度化を続けることが難しくなる。

## ランダウアーの原理から可逆計算へ

1961年、IBM の研究者ロルフ・ランダウアーは「情報の消去には必ずエネルギーが必要である」とする考え方を提唱した。これは現在ランダウアーの原理として知られている。たとえば NAND ゲートのような不可逆な論理操作では、出力から入力を一意に復元できないため、情報の消去が起こる。この過程ではエネルギーが消費され、発熱が避けられない。

1973年にはチャールズ・ベネットが、少なくとも理論上は情報を消去せずに計算する可逆計算が可能であることを示した。ここから、発熱を抑えた計算機の可能性が真剣に議論されるようになった。

## 量子計算という発想の登場

1980年、ポール・ベニオフは量子力学で記述される物理系を使えば熱を発生させずに計算できる可能性を示した。彼の着想は、微細化によって問題になる量子効果を、むしろ計算の資源として利用する方向への転換を意味していた。

その潜在能力を強く意識したのが、1982年のリチャード・ファインマンと、1985年のデイヴィッド・ドイッチュである。ファインマンは大規模な量子系のシミュレーションを古典コンピュータで行うことの難しさを指摘し、量子系をシミュレートするには量子コンピュータが自然だと考えた。ドイッチュは量子計算の理論的枠組みを与え、1989年には量子回路モデルを提案した。

## 量子アルゴリズムの発見と研究の拡大

1992年にはドイッチュとジョサが、古典アルゴリズムより有利な性質をもつ最初期の量子アルゴリズムを提示した。さらに 1994年にはピーター・ショアが素因数分解を高速に解く量子アルゴリズムを発表し、RSA 暗号の安全性に直接関わる可能性があることから、量子コンピュータ研究は一気に注目を集めた。1996年にはラブ・グローバーが探索問題に対する量子アルゴリズムを与え、量子計算の応用範囲はさらに広がった。

以後、量子ランダムウォーク、量子線形方程式ソルバー、量子化学計算、機械学習、金融モデリングなど、さまざまな方向で量子アルゴリズムの研究が進んだ。現在では、学術界だけでなく、大企業やスタートアップ、各国の国家プロジェクトが量子コンピュータ開発を支えている。

## まとめ

量子コンピュータ研究は、古典コンピュータの限界をどう乗り越えるかという問いから始まり、やがて量子力学そのものを計算資源として使うという発想に到達した。その後、ショアやグローバーのアルゴリズムの発見によって、量子計算は理論物理の話題にとどまらず、情報科学・工学・産業応用へと大きく広がっていった。
