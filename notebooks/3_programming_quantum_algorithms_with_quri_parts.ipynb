# 第3章 量子アルゴリズムの実行環境: QURI Parts を用いたプログラミング

本章では，量子アルゴリズムを実装・実行するための環境「QURI Parts」について学ぶ．まず，量子状態，量子ゲート，量子回路といった基本的な要素を QURI Parts でどのように表現・操作するかを説明する．続いて，オブザーバブルと呼ばれる量を定義してその期待値推定を行う方法や，変分パラメータを含む量子回路の扱い方を解説し，変分量子アルゴリズム（VQA）などへの応用を見据えたプログラミング方法を示す．章の後半では，QURI Parts を用いた応用例として，ベル不等式の代表例である CHSH 不等式の破れをシミュレーションし，量子もつれによる非局所的相関の理解を深める．
