# 第5章 変分量子回路に基づくアルゴリズム

この章では、変分量子回路を用いた量子・古典ハイブリッドアルゴリズムについて学んでいく。
量子コンピュータと古典コンピュータを組み合わせて計算を行う量子・古典ハイブリッドアルゴリズムは、NISQデバイスの活用法として重要であり、中でも変分量子回路を用いたものは盛んに研究されている。
この章で紹介するのは、NISQデバイスを実際の問題に応用していくために研究されてきたアルゴリズムばかりなので、基礎をしっかり身につけて、現状の問題点やさらなる発展の可能性を理解できるようになろう。
