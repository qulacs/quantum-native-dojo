# 第7章 量子位相推定アルゴリズムの応用

この章では、量子位相推定の応用先として、反復的位相推定、ショアのアルゴリズム、HHL アルゴリズムを扱う。現行の HHL 中心の構成から大きく再編する。

## 章構成（予定）

- 7-1. 反復的量子位相推定アルゴリズム
- 7-2. ショアのアルゴリズム（素因数分解）
- 7-3. HHL アルゴリズムを用いた連立一次方程式の解法
- コラム: 量子位相推定における補助ビットの削減手法
- コラム: 古典計算を行う量子回路と量子算術演算
- コラム: 量子ランダムアクセスメモリ（QRAM）

## 移行メモ

7-1 は現行資産の流用余地があるが、7-2 は新規比率が高い。章トップは書籍版の3節構成とコラム構成を先に確定させる。
