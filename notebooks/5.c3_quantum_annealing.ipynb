# コラム: 量子アニーリング

量子アニーリングは、組合せ最適化問題を量子系を用いて解くために考案された手法である。
QAOAと同様に、$n$ 桁の古典ビットに対する最適化問題の（近似）解 $z^*$ に対応する状態 $\ket{z}$ の測定確率（波動関数の振幅）が大きい $n$ 量子ビット状態を作成することで、最適化問題の解を探索する。
QAOA の変分量子状態 $\ket{\boldsymbol{\beta}, \boldsymbol{\gamma}}_p$ は量子アニーリングで用いられる状態から着想されたものであり、ある意味では量子アニーリングより広い範囲で量子状態を探索しているといえる。
このコラムでは簡単に量子アニーリングについて紹介する。

量子アニーリングの基礎になっているのは、量子力学における**断熱定理**である。
断熱定理とは、量子系が時間に依存するハミルトニアン $H(t)$ に従って**ゆっくりと**時間発展する場合に適用可能な定理である。
量子系の初期状態 $\ket{\psi(t=0)}$ が、時刻 $t=0$ のハミルトニアン $H(0)$ の基底状態であるとき、時間に依存するシュレーディンガー方程式

$$
i \pdv{t} \ket{\psi(t)} = H(t) \ket{\psi(t)}
$$

に従って時間発展した時刻 $t=T$ における量子状態 $\ket{\psi(t=T)}$ も、その時刻でのハミルトニアン $H(T)$ の**基底状態**になっている、というのが断熱定理の主張である。

解きたい組み合わせ最適化問題に対応する量子系のハミルトニアンを $H_\mathrm{prob}$ とする。
量子アニーリングでは、基底状態を簡単に用意できるハミルトニアン $H_\mathrm{init}$ から $H_\mathrm{prob}$ まで、量子系のハミルトニアンをゆっくりと変化させることで、$H_\mathrm{prob}$ の基底状態を得ることを狙う。
具体的には何らかの関数 $s(t)$ を用いて

$$
H(t) = (1 - s(t)) H_\mathrm{init} + s(t) H_\mathrm{prob}, \qquad s(0)=0, \ s(T)=1
$$

とハミルトニアンを時間変化させる。
断熱定理によれば、$t=0$ での初期状態が $H(0)=H_\mathrm{init}$ の基底状態なら、終時刻 $t=T$ での状態は $H(T)=H_\mathrm{prob}$ の基底状態になっているはずで、この状態を測定すれば組み合わせ最適化問題の解が得られるはずである。

QAOA に対応する例では、$H_\mathrm{init} = -\sum_{i=0}^{n-1} X_i$, $H_\mathrm{prob} = C(Z)$ と選ぶ。
そしてトロッター展開を用いて、時刻 $t_1, \cdots, t_p$ で時間発展演算子を近似した

$$
\ket{\psi(t=T)} \approx \prod_{i=1}^{p} e^{-iH(s(t_i))} \ket{\psi(t=0)}
$$

さらに

$$
\ket{\psi(t=T)} \approx \prod_{i=1}^{p} e^{-i(1-s(t_i))H_\mathrm{init}} e^{-is(t_i)H_\mathrm{prob}} \ket{\psi(t=0)}
$$

という状態を考える。
$H_\mathrm{init}$ の基底状態が $\ket{+}^{\otimes n} = \ket{s}$ であることと、$U_C(\gamma), U_X(\beta)$ の定義を思い出すと、状態 $\ket{\psi(t=T)}$ は QAOA の変分量子状態 $\ket{\boldsymbol{\beta}, \boldsymbol{\gamma}}_p$ においてパラメータを適当に選んだものであることがわかる。
断熱定理によって、$\ket{\psi(t=T)}$ は $H_\mathrm{prob}$ の基底状態に近いと予想されるから、QAOA はそのような状態を含んだ形でうまく変分量子回路を構成したアルゴリズムであるといえる。
