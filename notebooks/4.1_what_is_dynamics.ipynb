### 4.1.1 物理系のシミュレーション

古典・量子によらず，コンピュータの応用先として重要なものに，「物理系のダイナミクスのシミュレーション」がある．物理系のシミュレーションは，身のまわりの生活を成り立たせるうえで，なくてはならない技術である．例えば，ビルや航空機の設計，気象予報，創薬といった，ありとあらゆる場面でコンピュータを用いた物理系のシミュレーションが行われている．物理系のシミュレーションのゴールは，系の初期状態と，その系のダイナミクス，すなわち時間発展のルールが与えられたときに，一定時間後に系がどういう状態にあるかを知ることである。

例としてビリヤードのシミュレーションを考えよう。初期状態は各ボールの最初の位置と速度であり，時間発展のルールはニュートンの運動方程式である。

$$
m\frac{d}{dt}v = F
$$

ここで，$m$ はボールの質量，$v$ は速度，$F$ は力，$t$ は時間である。古典コンピュータによる数値シミュレーションでは，時間区間を小さな刻みに分け，たとえばオイラー法により

$$
v(t_j + \Delta t) = v(t_j) + \Delta t\left.\frac{dv}{dt}\right|_{t_j} + O(\Delta t^2) \approx v(t_j) + \frac{F(t_j)}{m}\Delta t
$$

のような近似式を反復的に用いて将来の状態を計算する。一般に，物理系のシミュレーションの本質は，時間発展を記述する微分方程式を数値的に解くことにある。

### 4.1.2 量子系のシミュレーション

量子系において，量子状態 $|\psi(t)\rangle$ の時間発展はシュレーディンガー方程式

$$
i\hbar\frac{d}{dt}|\psi(t)\rangle = H|\psi(t)\rangle
$$

によって記述される。$H$ はハミルトニアンというエルミート演算子であり，量子系の時間発展をつかさどる。量子コンピュータの文脈では，ハミルトニアンは巨大な行列だと考えてよいが，一般にはユニタリではないため，そのまま量子回路として実装することはできない点に注意が必要である。

ハミルトニアンは系のエネルギーに対応するオブザーバブルでもある。したがって，固有方程式

$$
H|\psi\rangle = E|\psi\rangle
$$

における固有値 $E$ は系がとりうるエネルギーを表し，対応する固有状態はエネルギー固有状態である。最も低いエネルギーに対応する状態は基底状態と呼ばれ，物質科学や量子化学においてとりわけ重要である。

ハミルトニアンが時間に依存しない場合，シュレーディンガー方程式の解は

$$
|\psi(t)\rangle = e^{-iHt/\hbar}|\psi(0)\rangle
$$

と書ける。したがって，量子ダイナミクスシミュレーションの目標は，この時間発展した状態や，そこから計算される物理量の期待値を求めることである。ここに現れるユニタリ演算子 $U=e^{-iHt/\hbar}$ を，量子回路としてどのように実装するかがアルゴリズムの中心課題になる。

量子系が大きくなると，古典コンピュータでその時間発展を厳密に扱うことは現実的でなくなる。$n$ 量子ビットの状態は $2^n$ 個の複素確率振幅をもつベクトルとして記述されるため，たとえば $n=100$ ではその情報量は天文学的に大きくなる。さらに，時間発展演算子を厳密に計算するには巨大行列の指数関数や対角化が必要になり，大規模な量子系では古典計算資源が急速に不足する。こうした背景が，量子コンピュータによる量子シミュレーションを重要な応用分野にしている。
