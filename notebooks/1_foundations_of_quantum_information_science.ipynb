# 第1章 量子情報科学の基礎

本章では，量子コンピュータの状態や操作を数学的に記述するために必要な，量子力学と量子情報科学の基礎について学んでいく．量子力学は一見とっつきにくい学問だが，量子コンピュータを扱うためのルールブックとして捉えた場合，意外にも理解が必須な事項は少ない．実際，量子力学の「運用規則」をマスターする，というくらいの気持ちで学ぶとちょうど良いので，あまり意気込まずそういうものなのかと思いながら読み進めてほしい．量子力学のルールを体になじませることが，Quantum Native に近づくための第一歩であるから，一緒にがんばろう．

本章の構成について述べる．まず，1.1 節と 1.2 節では，量子コンピュータが扱う情報の単位である量子ビットが 1 つだけの場合について，その記述の方法と操作を学ぶ．次に，1.3 節で複数の量子ビットがある場合について学ぶ．最後に 1.4 節で，量子ビットを操作する手順を表す量子回路というものを，どのように図示するかについて学び，量子コンピュータを理解するための基礎を完成させる．
