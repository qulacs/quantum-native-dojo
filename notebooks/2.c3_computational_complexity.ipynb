# コラム: 計算量について

ここで，古典コンピュータと量子コンピュータの計算量を比較する方法とその注意点について，量子フーリエ変換を例に見てみよう。古典コンピュータで離散フーリエ変換を素朴に計算すると $O(4^n)$，高速フーリエ変換を使っても $O(n2^n)$ の演算回数が必要になる。

一方，量子フーリエ変換では，量子回路中で使う 1 量子ビット・2量子ビットゲートの数に注目すると，必要なゲート数は $O(n^2)$ で済む。この比較だけを見ると，量子コンピュータが指数関数的に高速だと言いたくなる。

ただし，ここには重要な注意点がある。量子フーリエ変換の出力状態からすべての成分を高精度に読み出したい場合，測定を何度も繰り返す必要があり，その読み出しコストは一般に指数関数的に大きくなりうる。また，古典計算と量子計算では「1回の演算」の意味が異なるため，演算回数だけで実際の実行時間までそのまま比較できるわけではない。

したがって，「量子コンピュータの方が速い」という主張を見るときは，何を入力とし，何を出力として読み出したいのか，そして計算量をどのように定義して比較しているのかを丁寧に確認することが大切である。
