# 0-3. 量子コンピュータの代表的な方式

量子アルゴリズムが理論的に高速であっても、それを実際に実行するにはハードウェアが必要になる。量子ビットの実現方法は 1 つではなく、異なる物理系を用いた複数の方式が研究されている。この節では、まず量子コンピュータとして成立するための条件を整理し、その後に代表的な 3 方式である超伝導、イオントラップ、光方式を概観する。

## 0.3.1 ディヴィンチェンゾの基準

量子コンピュータは、単に量子ビットらしきものが存在するだけでは成立しない。デイヴィッド・ディヴィンチェンゾは、量子計算装置として意味をもつための基準として、次のような要件を整理した。

- よく特徴づけられた量子ビットをもち、拡張可能な物理システムであること
- 量子ビットを既知の基準状態に初期化できること
- 計算を終えるまで十分な寿命、すなわちコヒーレンス時間をもつこと
- 任意の量子アルゴリズムを実装できるユニバーサルゲートセットを実装できること
- 測定によって量子状態から古典的な情報を取り出せること

これらは、量子コンピュータが計算機として機能するための最低限の条件だと考えればよい。

## 0.3.2 超伝導

超伝導量子ビットは、極低温で電気抵抗がゼロになる超伝導体を用いた回路で量子ビットを実現する方式である。ジョセフソン接合を組み込んだ回路ではエネルギー準位が量子化され、そのうち 2 つの準位を $|0\rangle$ と $|1\rangle$ に対応させる。制御には主にマイクロ波パルスを用いる。

この方式の強みは、半導体や電子回路の製造技術を応用しやすく、大規模集積に向いている点にある。ゲート操作が高速で、企業による開発競争も活発である。一方で、極低温環境、大規模な配線、消費電力、チップサイズの増大など、スケールアップ時の技術課題も大きい。

## 0.3.3 イオントラップ

イオントラップ方式では、電子を 1 つ失ったイオンを電場や磁場で真空中に閉じ込め、それぞれを量子ビットとして使う。レーザー光を当てることで状態の操作や複数量子ビット間のゲートを実行する。

イオンは自然界の原子そのものであり、量子ビットごとのばらつきが小さい。そのため、非常に高精度なゲート操作と長いコヒーレンス時間が得られることが大きな利点である。一方で、レーザーや真空装置を含む光学系が大がかりになり、量子ビット数を大きく増やすことが難しいという課題がある。

## 0.3.4 光

光量子コンピュータは、光子を量子ビットとして使う方式である。光子は環境と相互作用しにくく、長距離を伝送しても情報を保ちやすいため、量子通信や量子ネットワークとの相性が良い。

この方式の魅力は、室温・常圧で扱える可能性があること、そして既存の光通信技術や光集積回路技術を活かしやすいことにある。その一方で、光子どうしの相互作用が弱く、2量子ビットゲートの効率的な実装が難しい。高性能な単一光子源や検出器も依然として重要な技術課題である。

## まとめ

超伝導、イオントラップ、光方式はいずれも有力だが、それぞれ強みと弱みが異なる。超伝導は高速性と集積性、イオントラップは高精度と長寿命、光は通信との親和性と常温動作の可能性が魅力である。現時点では単一の勝者が決まっているわけではなく、用途やスケールに応じて複数の方式が並行して発展していく可能性が高い。
